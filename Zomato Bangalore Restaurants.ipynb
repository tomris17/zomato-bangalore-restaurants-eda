# Zomato Bangalore Restaurants

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("zomato.csv")
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['url', 'phone'], axis=1, inplace=True, errors='ignore')
```

```python
df['ratings'] = df.rate.str.split('/', expand=True)[0].str.rstrip()
df['ratings'] = df['ratings'].str.replace('-', '-1')
df['ratings'].fillna('-1', inplace=True)
df['ratings'] = df['ratings'].str.replace('NEW', '-2')
df['ratings'] = pd.to_numeric(df['ratings'], errors='coerce')
df.drop('rate', axis=1, inplace=True, errors='ignore')
```

```python
if 'approx_cost(for two people)' in df.columns:
    df['approx_cost(for two people)'] = df['approx_cost(for two people)'].astype(str).str.replace(',', '')
    df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')
    df.rename(columns={'approx_cost(for two people)': 'cost_for_two'}, inplace=True)
```

```python
if 'rest_type' in df.columns:
    df['rest_type'] = df['rest_type'].astype(str).str.split(',', expand=True)[0].str.strip()
```

```python
df.to_csv("zomato_cleaned.csv", index=False)
```
